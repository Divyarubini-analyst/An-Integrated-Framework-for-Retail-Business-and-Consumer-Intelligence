```python
df=orders['Order_date'] = pd.to_datetime(orders['Order_date'])
```

```python
df
```

```python
date=orders['Order_date'].max() + pd.Timedelta(days=1)
```

```python
pddd=pd.merge(customers,orders,on="Customer_ID",how="right")
pddd
```

```python
order_items['Sales_Amount'] = order_items['Quantity'] * order_items['List_Price'] * (1 -order_items['Discount'])
```

```python
order_items
```

```python
tablejoin=pd.merge(order_items,pddd,on="Order_ID",how="right")
```

```python
tablejoin
```

```python
rfm = tablejoin.groupby('Customer_ID').agg(
    Recency=('Order_date', lambda x: (date - x.max()).days),
    Frequency=('Order_ID', 'nunique'),
    Monetary=('List_Price', 'sum')
).reset_index()
```

```python
rfm.head(10)
```

```python
rfm.to_sql(
    'customer_segments',
    con=engine,
    if_exists='replace',
    index=False)
print("RFM segmentation results exported successfully!")
```
